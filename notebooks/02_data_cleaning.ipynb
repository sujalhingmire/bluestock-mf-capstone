```python
import pandas as pd
import numpy as np
```

```python
fund_master = pd.read_csv("../data/raw/fund_master.csv")
nav = pd.read_csv("../data/raw/nav_history.csv")
performance = pd.read_csv("../data/raw/scheme_performance.csv")
holdings = pd.read_csv("../data/raw/portfolio_holdings.csv")
transactions = pd.read_csv("../data/raw/investor_transactions.csv")
sip = pd.read_csv("../data/raw/monthly_sip_inflows.csv")
category = pd.read_csv("../data/raw/category_inflows.csv")
aum = pd.read_csv("../data/raw/aum_by_fund_house.csv")
benchmark = pd.read_csv("../data/raw/benchmark_indices.csv")
folio = pd.read_csv("../data/raw/industry_folio_count.csv")
```

```python
datasets = {
    "Fund Master": fund_master,
    "NAV": nav,
    "Performance": performance,
    "Holdings": holdings,
    "Transactions": transactions,
    "SIP": sip,
    "Category": category,
    "AUM": aum,
    "Benchmark": benchmark,
    "Folio": folio
}

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.shape)
    print(df.isnull().sum().sum(), "missing values")
```

```python
fund_master = fund_master.drop_duplicates()
nav = nav.drop_duplicates()
performance = performance.drop_duplicates()
holdings = holdings.drop_duplicates()
transactions = transactions.drop_duplicates()
sip = sip.drop_duplicates()
category = category.drop_duplicates()
aum = aum.drop_duplicates()
benchmark = benchmark.drop_duplicates()
folio = folio.drop_duplicates()
```

```python
nav['date'] = pd.to_datetime(nav['date'])

transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date']
)

benchmark['date'] = pd.to_datetime(
    benchmark['date']
)
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
nav = nav[nav['nav'] > 0]

transactions = transactions[
    transactions['amount_inr'] > 0
]
```
